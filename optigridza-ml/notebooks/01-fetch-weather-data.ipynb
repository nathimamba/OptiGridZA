```python
import requests
import pandas as pd
import warnings
warnings.filterwarnings("ignore")

CITIES = {
    "Pretoria": (-25.7479, 28.2293),
    "Cape Town": (-33.9249, 18.4241),
    "Durban": (-29.8587, 31.0218),
}

frames = []
for city, (lat, lon) in CITIES.items():
    url = (
        f"https://archive-api.open-meteo.com/v1/archive"
        f"?latitude={lat}&longitude={lon}"
        f"&start_date=2024-01-01&end_date=2025-12-31"
        f"&hourly=shortwave_radiation,cloud_cover,temperature_2m"
        f"&timezone=Africa%2FJohannesburg"
    )
    r = requests.get(url).json()
    df = pd.DataFrame(r["hourly"])
    df["city"] = city
    frames.append(df)

weather = pd.concat(frames, ignore_index=True)
weather.to_csv("../data/raw/historical_weather.csv", index=False)
print(weather.shape)
weather.head(50)
```
